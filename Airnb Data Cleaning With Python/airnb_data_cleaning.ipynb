```python
columns_to_keep = ['id', 'NAME', 'host id', 'host_identity_verified', 'host name',
       'neighbourhood group', 'neighbourhood', 'lat', 'long', 'country',
       'country code', 'instant_bookable', 'cancellation_policy', 'room type',
       'Construction year', 'price', 'service fee', 'minimum nights',
       'number of reviews', 'last review']


```

```python
columns_to_drop = ['reviews per month','review rate number', 'calculated host listings count',
       'availability 365', 'house_rules', 'license']
```

### Filter Method

```python
filter_data = data[columns_to_keep]
```

### Renaming The Column

```python
filter_data.columns
```

```python
new_column_names = []

for i in filter_data.columns:
    new_column_names.append(i.capitalize())
```

```python
filter_data.columns = new_column_names
```

```python
filter_data.duplicated().sum()
```

```python
filter_data.drop_duplicates(inplace= True)
```

```python
missing_data = filter_data.isna().sum()
```

```python
missing_data/ len(filter_data) * 100
```

```python
filter_data.info()
```

```python
remove_null_data = filter_data.dropna()
```

```python
remove_null_data.info()
```

```python
remove_null_data['Host_identity_verified'] = remove_null_data['Host_identity_verified'].str.strip().str.capitalize()
```

```python
remove_null_data.head()
```

```python
remove_null_data['Neighbourhood'].head(20)
```

```python
remove_null_data['Cancellation_policy'].str.strip().str.capitalize()
```

```python
remove_null_data['Cancellation_policy'] = remove_null_data.loc[:,'Cancellation_policy'].str.strip().str.capitalize()
```

```python
remove_null_data.head()
```

```python
remove_null_data['Price'] = remove_null_data['Price'].str.replace('$','')
```

```python
remove_null_data['Service fee'] = remove_null_data.loc[:,'Service fee'].str.replace('$', '')
```

```python
remove_null_data.loc[:,'Country code'] = remove_null_data.loc[:,'Country code'].str.strip()
```

```python
final_stage_data = remove_null_data.copy()
```

```python
final_stage_data.head()
```

```python
final_stage_data.info()
```

```python
final_stage_data['Price'] = final_stage_data['Price'].str.replace(',','')
```

```python
final_stage_data['Price'] = pd.to_numeric(final_stage_data['Price'])
```

```python
final_stage_data['Service fee'] = pd.to_numeric(final_stage_data['Service fee'])
```

```python
final_stage_data.info()
```

```python
final_stage_data.head(10)
```

```python
final_stage_data['Construction year'] = final_stage_data['Construction year'].astype(int)
```
